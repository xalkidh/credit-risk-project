```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/credit_default_clean.csv")
print(df.shape)
df.head()
```

```python
for col in ["SEX", "EDUCATION", "MARRIAGE"]:
    print(df.groupby(col)["DEFAULT"].agg(["count", "mean"]).round(3))
    print()
```

```python
print(df.groupby("PAY_1")["DEFAULT"].agg(["count", "mean"]).round(3))
```

```python
# Correlation heatmap of all numeric columns (Spearman suits the ordinal PAY columns)
corr = df.drop(columns=["SEX"]).corr(method="spearman")

# Hide the upper triangle to avoid duplicated information
mask = np.triu(np.ones_like(corr, dtype=bool))

fig, ax = plt.subplots(figsize=(14, 11))
sns.heatmap(corr, mask=mask, cmap="coolwarm", center=0, vmin=-1, vmax=1,
            square=True, linewidths=0.3, cbar_kws={"shrink": 0.7}, ax=ax)
ax.set_title("Spearman correlation matrix")
plt.tight_layout()
plt.show()

# Strongest correlations with the target
print(corr["DEFAULT"].drop("DEFAULT").sort_values(key=abs, ascending=False).round(3).head(8))
```
